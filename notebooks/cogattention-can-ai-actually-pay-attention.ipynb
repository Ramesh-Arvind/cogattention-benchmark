# 🧠 CogAttention: Can AI Actually Pay Attention?
## Adapting 70 Years of Cognitive Psychology to Benchmark LLMs

> **We don't test what models know. We test how they process.**

Most benchmarks measure knowledge or reasoning. CogAttention measures **attention** — the cognitive ability to focus, filter, sustain, shift, and notice. Grounded in paradigms from Stroop (1935) to the Gorilla Experiment (1999).

**Try it yourself below** — see if you can beat the AI.

```python
from IPython.display import HTML, display
import random, time

display(HTML('''
<div style="background: linear-gradient(135deg, #1a1a2e 0%, #16213e 100%); 
     padding: 30px; border-radius: 15px; border: 2px solid #e94560; margin: 10px 0;">
  <h2 style="color: #e94560; margin-top: 0;">🎯 Try It: The Stroop Test</h2>
  <p style="color: #aaa;">What <b>COLOR</b> is each word printed in? (Ignore what the word says!)</p>
  <div style="display: flex; gap: 30px; justify-content: center; padding: 20px; flex-wrap: wrap;">
    <span style="font-size: 42px; font-weight: bold; color: #0000FF;">RED</span>
    <span style="font-size: 42px; font-weight: bold; color: #FF0000;">GREEN</span>
    <span style="font-size: 42px; font-weight: bold; color: #008000;">YELLOW</span>
    <span style="font-size: 42px; font-weight: bold; color: #FFD700;">BLUE</span>
  </div>
  <p style="color: #666; font-size: 13px; text-align: center;">
    Answers: Blue, Red, Green, Yellow (Gold) — If you said "Red, Green, Yellow, Blue" you read the <i>words</i>, not the <i>colors</i>.<br>
    This is the <b>Stroop Effect</b> (1935). Your brain automatically reads the word, interfering with naming the ink color.<br>
    <b>LLMs show the exact same interference.</b> CogAttention measures it.
  </p>
</div>
'''))
```

## 🔄 Try It: Can You Track the Swaps?

Three people each hold an item:
- **Alice** holds a 🔑 red key
- **Bob** holds a 📘 blue book  
- **Carol** holds a 🪙 green coin

Now follow the swaps:
1. Alice and Bob swap items.
2. Bob and Carol swap items.
3. Alice and Carol swap items.

**Who holds what now?** Think about it before running the cell below.


```python
from IPython.display import HTML, display
import ipywidgets as widgets

# The answer
answer_html = '''
<div style="background: #16213e; padding: 20px; border-radius: 10px; border: 1px solid #53d8a8; margin: 10px 0;">
  <h4 style="color: #53d8a8;">Answer:</h4>
  <table style="color: #ccc;">
    <tr><td style="padding: 5px 15px;">Start:</td><td>Alice=🔑  Bob=📘  Carol=🪙</td></tr>
    <tr><td style="padding: 5px 15px;">Swap 1 (A↔B):</td><td>Alice=📘  Bob=🔑  Carol=🪙</td></tr>
    <tr><td style="padding: 5px 15px;">Swap 2 (B↔C):</td><td>Alice=📘  Bob=🪙  Carol=🔑</td></tr>
    <tr><td style="padding: 5px 15px;">Swap 3 (A↔C):</td><td>Alice=🔑  Bob=🪙  Carol=📘</td></tr>
  </table>
  <p style="color: #e94560; margin-top: 10px;">
    <b>This was Easy mode (3 people, 3 swaps).</b><br>
    At Frontier difficulty, it's <b>5 people, 12 swaps</b>. 
    Qwen-72B drops from 1.0 accuracy to 0.19. Even 72 billion parameters can't reliably track that many objects.
  </p>
</div>
'''

btn = widgets.Button(description="Reveal Answer", button_style='info',
                     layout=widgets.Layout(width='200px', height='40px'))
output = widgets.Output()

def on_click(b):
    with output:
        output.clear_output()
        display(HTML(answer_html))

btn.on_click(on_click)
display(btn, output)
```

## 👀 Try It: Can You Spot the Anomaly?

**Your task: Count how many times the word "the" appears in this passage.**

> The quarterly report confirmed that the revenue exceeded projections by a comfortable margin. The board reviewed the findings and noted that the expansion into new markets had been successful. The team in Oslo presented the latest figures, which showed the growth rate holding steady. Le soleil se couchait derrière les montagnes, peignant le ciel en nuances d'or et de pourpre. The infrastructure upgrades were completed ahead of schedule, and the client satisfaction scores reached the highest point in the company's history.

**How many times does "the" appear? Did you notice anything unusual?**


```python
from IPython.display import HTML, display
import ipywidgets as widgets

answer_html = '''
<div style="background: #16213e; padding: 20px; border-radius: 10px; border: 1px solid #9b59b6; margin: 10px 0;">
  <h4 style="color: #9b59b6;">Answer:</h4>
  <p style="color: #ccc;"><b>"the" appears 11 times.</b></p>
  <p style="color: #e94560;"><b>The anomaly:</b> There's a sentence in <i>French</i> in the middle of an English business report!<br>
  <i>"Le soleil se couchait derrière les montagnes..."</i> (The sun was setting behind the mountains...)</p>
  <p style="color: #aaa; font-size: 13px; margin-top: 10px;">
    This is <b>Inattentional Blindness</b> (Simons & Chabris, 1999) — when you're focused on counting,
    you miss the gorilla walking through the scene.<br><br>
    <b>Phi-3.5 scores 0.16 on this task</b> — it almost never notices the anomaly.<br>
    <b>Humans score 0.68</b> — we're much better at noticing the unexpected, even under load.
  </p>
</div>
'''

btn = widgets.Button(description="Reveal Answer", button_style='warning',
                     layout=widgets.Layout(width='200px', height='40px'))
output = widgets.Output()

def on_click(b):
    with output:
        output.clear_output()
        display(HTML(answer_html))

btn.on_click(on_click)
display(btn, output)
```

## 📊 How Do Frontier Models Perform?


```python
import plotly.graph_objects as go

abilities = ['Capacity', 'Sustained', 'Selective', 'Shifting', 'Stimulus-Driven']

profiles = {
    'Qwen-72B': [0.87, 0.82, 0.91, 0.85, 0.48],
    'Human (n=25)': [0.81, 0.79, 0.85, 0.63, 0.68],
}

colors = {'Qwen-72B': '#4ec9b0', 'Human (n=25)': '#e94560'}

fig = go.Figure()
for name, vals in profiles.items():
    fig.add_trace(go.Scatterpolar(
        r=vals + [vals[0]], theta=abilities + [abilities[0]],
        fill='toself', name=name, opacity=0.3,
        line=dict(color=colors[name], width=3),
    ))

fig.update_layout(
    polar=dict(
        bgcolor='#16213e',
        radialaxis=dict(visible=True, range=[0, 1], gridcolor='#333355', tickfont=dict(color='#aaa')),
        angularaxis=dict(gridcolor='#333355', tickfont=dict(color='#ccc', size=14)),
    ),
    paper_bgcolor='#1a1a2e',
    font=dict(color='#ccccdd'),
    title=dict(text='Cognitive Profile: Human vs Best LLM', font=dict(size=18, color='#e94560')),
    legend=dict(bgcolor='rgba(22,33,62,0.8)', bordercolor='#444466', font=dict(size=13)),
    height=500,
)
fig.show()
```


**The shapes tell the story.** Humans (red) are better at noticing the unexpected. LLMs (green) are better at brute-force tracking and rule-following. They have **inverted cognitive profiles** — reflecting fundamentally different architectures.

The full benchmark tests 16 task types across 860 items with 5 difficulty tiers, including multimodal Visual Stroop for VLMs.
