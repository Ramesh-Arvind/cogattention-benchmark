# 🧠 CogAttention: Can AI Actually Pay Attention?
## Adapting 70 Years of Cognitive Psychology to Benchmark LLMs

> **We don't test what models know. We test how they process.**

Most benchmarks measure knowledge. CogAttention measures **attention** — the ability to focus, filter, sustain, shift, and notice the unexpected. Every task is grounded in a classic cognitive psychology paradigm.

**Try the demos below** — see if you can beat the AI.

## 📊 The Big Finding: Inverted Cognitive Profiles


```python
import plotly.graph_objects as go

abilities = ['Capacity', 'Sustained', 'Selective', 'Shifting', 'Stimulus-Driven']

fig = go.Figure()
fig.add_trace(go.Scatterpolar(
    r=[0.87, 0.82, 0.91, 0.85, 0.48, 0.87], theta=abilities + [abilities[0]],
    fill='toself', name='Best LLM (Qwen-72B)', opacity=0.3,
    line=dict(color='#4ec9b0', width=3),
))
fig.add_trace(go.Scatterpolar(
    r=[0.81, 0.79, 0.85, 0.63, 0.68, 0.81], theta=abilities + [abilities[0]],
    fill='toself', name='Human (n=25)', opacity=0.3,
    line=dict(color='#e94560', width=3),
))

fig.update_layout(
    polar=dict(
        bgcolor='#16213e',
        radialaxis=dict(visible=True, range=[0, 1], gridcolor='#333355', tickfont=dict(color='#aaa')),
        angularaxis=dict(gridcolor='#333355', tickfont=dict(color='#ccc', size=14)),
    ),
    paper_bgcolor='#1a1a2e', font=dict(color='#ccccdd'),
    title=dict(text='Human vs Best LLM: Who Pays Attention Better?', font=dict(size=18, color='#e94560')),
    legend=dict(bgcolor='rgba(22,33,62,0.8)', bordercolor='#444466', font=dict(size=13)),
    height=500,
)
fig.show()
```


Humans are better at **noticing the unexpected** (Stimulus-Driven). LLMs are better at **brute-force tracking** (Capacity, Shifting). They have opposite strengths — reflecting fundamentally different architectures.

The full benchmark tests 16 task types across 860 items with 5 difficulty tiers.
